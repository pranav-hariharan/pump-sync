```python
import pandas as pd
print(pd.read_csv("../data/raw/songs.csv", nrows=0).columns.tolist())
```

```python
cols = ["id", "name", "artists", "year", "genre", "popularity",
        "danceability", "energy", "valence", "key", "loudness", "mode",
        "speechiness", "acousticness", "instrumentalness", "liveness",
        "tempo", "duration_ms", "total_artist_followers", "avg_artist_popularity"]
new = pd.read_csv("../data/raw/songs.csv", usecols=cols)
print(new.shape)
print(new["year"].describe())
print(new["year"].value_counts().sort_index().tail(12))
print(new["genre"].value_counts().head(25))
```

```python
queries = ["POWER", "FE!N", "Praise The Lord", "STARGAZING", "METAMORPHOSIS",
           "Stop Breathing", "Lord Pretty Flacko", "Stick", "MURDER IN MY MIND",
           "NEON BLADE", "RAVE", "Fighting My Demons", "Till I Collapse",
           "X Gon", "Can't Be Touched"]

for q in queries:
    hits = new[new["name"].str.contains(q, case=False, regex=False)]
    print(q, "->", len(hits), hits[["name", "artists", "year"]].head(2).values.tolist())
```

```python
hh = new[(new["genre"] == "Hip-Hop") & (new["year"] >= 2015)
         & (new["energy"] > 0.75) & (new["popularity"] >= 50)]
print(len(hh))
hh.sample(30, random_state=1)[["name", "artists", "year", "popularity"]]
```
